```python
import tiktoken
import os
from dotenv import load_dotenv
from google import genai

load_dotenv()

client = genai.Client()

# Cell 2: Tokenization Inspection
encoding = tiktoken.get_encoding("cl100k_base")
sample_text = "Bachelor of Computer Engineering (BCT)"
tokens = encoding.encode(sample_text)
print(f"Text: {sample_text}")
print(f"Token Count: {len(tokens)}")

# Cell 3: Context-Grounded Prompt Execution
res = client.models.generate_content(
    model="gemini-3.5-flash",
    contents="Briefly define the context window in LLMs."
)
print(res.text)
```
