```python
#calling api directlty
url = "https://catfact.ninja/fact"

response = requests.get(url)

# Check if successful
if response.status_code == 200:
    data = response.json()  # Converting JSON response to Python dictionary
    print("API Response:")
    print(data["fact"])
else:
    print(f"Failed to fetch data. Status code: {response.status_code}")
```

```python
#calling api using file open method
import requests

with open(".env", "r") as file:
    for line in file:
        line = line.strip()
        if "API_KEY=" in line:
            # Extract everything after "API_KEY="
            API_KEY = line.split("=", 1)[1].strip(" \"'")
            break
        
url = "http://api.weatherapi.com/v1/current.json"

query_params = {
    "key": API_KEY,
    "q": "Kathmandu",
    "aqi": "no"
}

response = requests.get(url, params=query_params)

if response.status_code == 200:
    data = response.json()
    print(f"Weather in {data['location']['name']}, {data['location']['country']}:")
    print(f"Temperature: {data['current']['temp_c']}°C")
    print(f"Condition: {data['current']['condition']['text']}")
else:
    print(f"Error {response.status_code}: {response.text}")
```

```python
#callibng the weather api using requests library
import os
import requests

# function to load API_KEY into system environment
def load_dotenv(env_path=".env"):
    if os.path.exists(env_path):
        with open(env_path, "r", encoding="utf-8") as file:
            for line in file:
                if "=" in line and not line.strip().startswith("#"):
                    k, v = line.strip().split("=", 1)
                    os.environ.setdefault(k.strip(), v.strip().strip("\"'"))

# load api key from .env file
load_dotenv()
API_KEY = os.getenv("API_KEY")

if not API_KEY:
    raise ValueError("API_KEY missing in .env file!") #shows error in loading api key

# fetch current weather for Kathmandu
url = "http://api.weatherapi.com/v1/current.json"
params = {"key": API_KEY, "q": "Kathmandu", "aqi": "no"}

response = requests.get(url, params=params, timeout=5)

# check success & show weather output
if response.ok:
    print(response.json())
else:
    print(f"Error {response.status_code}: {response.text}")
```

```python
#payload api call
import json
import requests

url = "https://jsonplaceholder.typicode.com/posts"

payload = {
    "user_id": 101,
    "location": "Kathmandu",
    "status": "active"
}

try:
    response = requests.post(url, json=payload, timeout=5)
    
    if response.status_code in [200, 201]:
        data = response.json()
        print("Success! Server received:\n")
        print(json.dumps(data, indent=4))
        
    else:
        print(f"Server problem! Status code: {response.status_code}")

except requests.exceptions.RequestException:
    print("Could not connect to the server or website is down.")
```
