```python
# WRITE mode 
with open("notes.txt", "w") as file:
    file.write("Hello! This is my first text file in Python.")

print("File created successfully!")
```

```python
# READ mode
with open("notes.txt", "r") as file:
    content = file.read()

print("File Content:")
print(content)
```

```python
# APPEND mode
with open("notes.txt", "a") as file:
    file.write("\nThis is an additional line.")

# READ mode
with open("notes.txt", "r") as file:
    content = file.read()

print("File SUCCESSFULLY UPDATED!")
print("Updated File Content:")
print(content)
```

```python
import json

# Python dictionary
student = {
    "name": "Aastha",
    "age": 22,
    "skills": ["Python", "Database", "Git"]
}

# Save dictionary directly into a .json file
with open("student.json", "w") as file:
    json.dump(student, file, indent=4)  # indent=4 makes it neatly formatted line by line 

print("JSON file saved!")
```

```python
import json

# Read data from the JSON file
with open("student.json", "r") as file:
    data = json.load(file)  # Loads JSON data back into a Python dictionary

print("Loaded Student Name:", data["name"])
print("First Skill:", data["skills"][0])
```

```python
import json

# reading existing data from student.json
with open("student.json", "r") as file:
    data = json.load(file)

# STEP 2:add in the list in Python
data["skills"].append("React")
data["skills"].append("Docker")

# writing back to student.json
with open("student.json", "w") as file:
    json.dump(data, file, indent=4)

print("Updated skills successfully!")
```
