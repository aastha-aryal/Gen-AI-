# Exercise 3: Find Sum, Average, Maximum, and Minimum from a List

-Print all marks
-Find total marks
-Find average marks
-Find highest mark
-Find lowest mark

```python
subjects = [
    "Python",
    "Database",
    "AI",
    "Networks",
    "Software Engineering"
]
for subject in subjects:
    print("Subject:", subject)
```

```python
# Exercise 3: List Calculations

marks = [85, 90, 78, 92, 88, 20, 62, 75, 80, 95, 98]

print("Marks:")
print(marks)

total = sum(marks)
average = total / len(marks)
highest = max(marks)
lowest = min(marks)

print("Total marks:", total)
print("Average marks:", average)
print("Highest mark:", highest)
print("Lowest mark:", lowest)
```

```python
# Exercise 3: Student Result Analysis

student_name = "Aastha"

subjects = {
    "Python": 85,
    "Database": 90,
    "Artificial Intelligence": 78,
    "Computer Networks": 92,
    "Software Engineering": 88
}


print("Student Name:", student_name)

print("\nSubject-wise Marks:")

for subject, mark in subjects.items():
    print(subject, ":", mark)


# Total and Average

total_marks = sum(subjects.values())

average_marks = total_marks / len(subjects)


# Highest and Lowest marks

highest_subject = max(subjects, key=subjects.get)

lowest_subject = min(subjects, key=subjects.get)


print("\nResult Summary")
print("Total Marks:", total_marks)
print("Average Marks:", average_marks)

print("Highest Marks is", subjects[highest_subject], "in", highest_subject)
print("Lowest Marks is", subjects[lowest_subject], "in", lowest_subject)
```

```python
# Exercise 9: Student Marks using while loop

marks = [85, 90, 78, 92, 88]

student_count = 0

while student_count < len(marks):
    print(f"Student {student_count + 1} Marks: {marks[student_count]}")
    student_count = student_count + 1


total_marks = sum(marks)
average_marks = total_marks / len(marks)

print("\nTotal Marks:", total_marks)
print("Average Marks:", average_marks)
print("Highest Marks:", max(marks))
print("Lowest Marks:", min(marks))
```

```python
#once you find the target item exiting the loop
users = ["Alice", "Bob", "Charlie", "David"]
for user in users:
    if user == "Charlie":
        print("User found!")
        break  # Exit loop immediately
    print("Checking:", user)
```

```python
#skip missing entries in a list
scores = [85, None, 92, 45, None, 78]
for score in scores:
    if score is None:
        continue  # Skip missing entries
    print("Processing valid score:", score)
```

```python
#if....else condition in a loop 
numbers = [1,3, 5, 7, 9]
for num in numbers:
    if num % 2 != 0:
        print("Found an odd number!")
        break
else:
    print("All numbers are even.")  # Runs only if no 'break' occurred
```
