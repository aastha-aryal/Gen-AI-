Exercise 4- Function definig


```python
# 1. DEFINE the function
def greet():
    print("Hello, welcome to Python learning!")

# 2. CALL (run) the function
greet()
```

```python
#create a var outside the function
x = "awesome"
def myfunc():
    print("Python is " + x)

myfunc()
```

```python
def double_number(num):
    print(num * 2)

# Call it:
double_number(5)
double_number(10)
```

```python
def add_numbers(a, b):
    return a + b

# Call it and save the result:
result = add_numbers(3, 7)
print("Result is:", result)
```

```python
def check_even(num):
    if num % 2 == 0:
        return "Even"
    else:
        return "Odd"

print(check_even(4))
print(check_even(7))
```

```python
def get_average(marks):
    total = sum(marks)
    count = len(marks)
    return total / count

my_marks = [80, 90, 100]
print("Average:", get_average(my_marks))
```

```python
def get_highest(scores):
    # max() on dictionary values gets the highest number
    return max(scores.values())

marks_dict = {"Python": 85, "Database": 92, "AI": 78}
print("Highest Mark:", get_highest(marks_dict))
```

```python
# defining the function
def is_passing(score):
    if score >= 40:
        return "Pass"
    else:
        return "Fail"
# calling the function
print(is_passing(85))  # Should print: Pass
print(is_passing(32))  # Should print: Fail
```
