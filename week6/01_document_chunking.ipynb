# Week 6 - 01: Document Chunking




```python
document = """
Artificial Intelligence is a field of computer science.
Machine learning allows computers to learn patterns from data.
Natural language processing helps computers understand human language.
Vector databases store numerical representations called embeddings.
Semantic search finds text based on meaning instead of only exact keywords.
"""

print(document)
```

```python
# split() separates the text wherever it finds whitespace.
words = document.split()

print(words)
print("Number of words:", len(words))
```

```python
# Simple fixed-size chunking:12 words

chunk_size = 12
chunks = []

for start in range(0, len(words), chunk_size):              # range() creates the positions we need.
    chunk = " ".join(words[start:start + chunk_size])       # [start:end] selects part of a list.
    chunks.append(chunk)

print("Number of chunks:", len(chunks))

for i, chunk in enumerate(chunks):
    print(f"Chunk {i}: {chunk}")
```

## A slightly better version: character chunks

In real projects, chunking is often more careful. For learning, we can use a simple character limit.

The `overlap` keeps a few characters from the previous chunk. This helps avoid losing meaning at the boundary.


```python
def chunk_text(text, chunk_size=200, overlap=40):
    # This function returns a list of smaller text pieces.

    chunks = []
    start = 0

    while start < len(text):
        end = start + chunk_size
        chunk = text[start:end].strip()

        if chunk:
            chunks.append(chunk)

        # Move forward, but keep some overlap with the previous chunk.
        start = end - overlap

    return chunks


text_chunks = chunk_text(document, chunk_size=120, overlap=20)

for i, chunk in enumerate(text_chunks):
    print(f"\n--- Chunk {i} ---")
    print(chunk)
```

## What you learned

- `split()` turns text into a list of words.
- `join()` combines words back into text.
- A `for` loop helps create many chunks.
- A function lets us reuse chunking logic.
- `overlap` means neighboring chunks share a small amount of text.
